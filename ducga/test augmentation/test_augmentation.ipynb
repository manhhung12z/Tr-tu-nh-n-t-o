```python
import cv2
import matplotlib.pyplot as plt
import albumentations as A
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

def letterbox_simulation(image_path, target_size=(640, 640), pad_color=(114, 114, 114)):
    # 1. Đọc ảnh gốc và chuyển sang RGB
    img = cv2.imread(image_path)
    img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
    h_orig, w_orig = img.shape[:2]
    th, tw = target_size

    # 2. Tính toán tỷ lệ co phóng (scale ratio)
    # Chọn tỷ lệ nhỏ hơn để đảm bảo cả 2 cạnh đều vừa vặn bên trong kích thước đích
    r = min(tw / w_orig, th / h_orig)

    # Kích thước mới của ảnh sau khi resize (vẫn giữ nguyên tỷ lệ dài-rộng)
    new_w = int(round(w_orig * r))
    new_h = int(round(h_orig * r))

    # Tiến hành resize ảnh theo kích thước mới
    img_resized = cv2.resize(img, (new_w, new_h), interpolation=cv2.INTER_LINEAR)

    # 3. Tính toán khoảng đệm (Padding) cần thêm vào các cạnh
    # Chia đôi khoảng thừa để ảnh nằm chính giữa khung hình
    pad_w = (tw - new_w) / 2
    pad_h = (th - new_h) / 2

    top, bottom = int(round(pad_h - 0.1)), int(round(pad_h + 0.1))
    left, right = int(round(pad_w - 0.1)), int(round(pad_w + 0.1))

    # Tạo ảnh Letterbox bằng cách thêm viền đệm màu xám xung quanh
    letterbox_img = cv2.copyMakeBorder(
        img_resized, top, bottom, left, right, 
        borderType=cv2.BORDER_CONSTANT, value=pad_color
    )
    
    return img, letterbox_img

# --- CHẠY MÔ PHỎNG VÀ VẼ ĐỒ THỊ ---
# Thay bằng đường dẫn ảnh lá cây thực tế trong tập dữ liệu (nên chọn ảnh hình chữ nhật dài hoặc dẹt)
image_path = "4-dogwood_pm_jpg.rf.7890a89380c4c11f354b144c9e0ef31e.jpg" 

try:
    orig, letterboxed = letterbox_simulation(image_path, target_size=(640, 640))

    # Vẽ hai ảnh cạnh nhau để so sánh đưa vào báo cáo
    fig, axes = plt.subplots(1, 2, figsize=(14, 7))

    # Ảnh gốc hiển thị đúng kích thước thực tế
    axes[0].imshow(orig)
    axes[0].set_title(f"Ảnh gốc ban đầu\nKích thước: {orig.shape[1]}x{orig.shape[0]}", fontsize=12, fontweight='bold')
    axes[0].axis("on") # Bật trục tọa độ để thấy rõ kích thước bị thay đổi

    # Ảnh sau khi xử lý Letterbox
    axes[1].imshow(letterboxed)
    axes[1].set_title(f"Ảnh sau khi xử lý Letterbox\nKích thước chuẩn: {letterboxed.shape[1]}x{letterboxed.shape[0]}", fontsize=12, fontweight='bold')
    axes[1].axis("on")

    plt.tight_layout()
    plt.savefig("mo_phong_letterbox.png", dpi=300)
    plt.show()

except Exception as e:
    print("Vui lòng kiểm tra lại đường dẫn file ảnh của bạn:", e)
```
